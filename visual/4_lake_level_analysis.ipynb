#### Get the date of minimal lake level and maximal lake level.

```python
def get_date_min_max(heights_lake):
    id_min = chenghai_wse['height_weighted_mean'].argmin()
    id_max = chenghai_wse['height_weighted_mean'].argmax()
    date_min = chenghai_wse['date'][id_min]
    date_max = chenghai_wse['date'][id_max]
    return date_min, date_max

```

```python
date_min_chenghai, date_max_chenghai = get_date_min_max(chenghai_wse['height_weighted_mean'])
date_min_dianchi, date_max_dianchi = get_date_min_max(dianchi_wse['height_weighted_mean'])
date_min_erhai, date_max_erhai = get_date_min_max(erhai_wse['height_weighted_mean'])
date_min_fuxian, date_max_fuxian = get_date_min_max(fuxian_wse['height_weighted_mean'])
date_min_lugu, date_max_lugu = get_date_min_max(lugu_wse['height_weighted_mean'])
date_min_qilu, date_max_qilu = get_date_min_max(qilu_wse['height_weighted_mean'])
date_min_xingyun, date_max_xingyun = get_date_min_max(xingyun_wse['height_weighted_mean'])
date_min_yangzonghai, date_max_yangzonghai = get_date_min_max(yangzonghai_wse['height_weighted_mean'])
date_min_yilong, date_max_yilong = get_date_min_max(yilong_wse['height_weighted_mean'])
print('date_min:', date_min_chenghai,date_min_dianchi,date_min_erhai,date_min_fuxian,date_min_lugu,date_min_qilu,date_min_xingyun,date_min_yangzonghai,date_min_yilong)
print('date_max:', date_max_chenghai,date_max_dianchi,date_max_erhai,date_max_fuxian,date_max_lugu,date_max_qilu,date_max_xingyun,date_max_yangzonghai,date_max_yilong)

```

```python
def plot_lake_heights(heights_lake, date_ticks, ax, y_var = 'height_weighted_mean', uncertain_var = None,
                      h_range=None, markersize=6, h_color='blue', h_label='Lake level'):
      dates_decimal = [date_to_decimal(date+'T00:00:00') for date in date_ticks]
      if uncertain_var is not None:
            ax.fill_between(heights_lake['date'], 
                  np.array([*heights_lake[y_var]]) - np.array([*heights_lake[uncertain_var]]), \
                  np.array([*heights_lake[y_var]]) + np.array([*heights_lake[uncertain_var]]), 
                  color='Brown', alpha=0.1)
      ax.plot(heights_lake['date'], heights_lake[y_var], '--.', 
            color=h_color, linewidth=0.6, markersize=markersize, label=h_label)
      ax.set_xticks(dates_decimal); ax.set_xticklabels(date_ticks, rotation=0)
      ax.legend(loc='upper right', fontsize=9)
      if h_range is not None:
            ax.set_ylim([h_range[0], h_range[1]]); 
            ax.set_yticks([h_range[0]+0.1+i*0.4 for i in range(0, 4)]);
```

```python
dates_ticks = ['2023-08-15','2024-01-15', '2024-06-15', '2024-11-15']
font_config = {'fontsize': 11, 'fontweight': 'bold', 'color':'FireBrick'}
fig, ax = plt.subplots(1,3,figsize=(10, 2.1), constrained_layout=True)
# fig.subplots_adjust(wspace=0.3)
## Chenghai
plot_lake_heights(heights_lake=chenghai_wse, markersize=7, y_var='height_weighted_mean', 
                  uncertain_var='height_weighted_dev', date_ticks=dates_ticks, h_range=[1495.8, 1497.45], ax=ax[0])
# ax[0].text(0.62, 0.1, f"Range: {max_range_chenghai:.2f}m", 
                # transform=ax[0].transAxes, fontdict=font_config)
# ax[0].set_title('Chenghai', fontsize=12, x=0.47, y=0.83, fontweight='bold')
print(round(max_range_chenghai, 2))
## Dianchi
plot_lake_heights(heights_lake=dianchi_wse, markersize=7, y_var='height_weighted_mean', 
                  uncertain_var='height_weighted_dev', date_ticks=dates_ticks, h_range=[1887.2, 1888.4], ax=ax[1])
ax[1].set_ylim([1887.3, 1888.2])
# ax[1].text(0.62, 0.1, f"Range: {max_range_dianchi:.2f}m", 
#                 transform=ax[1].transAxes, fontdict=font_config)
# ax[1].set_title('Dianchi', fontsize=12, y=0.83, fontweight='bold')
print(round(max_range_dianchi, 2))
## Erhai
plot_lake_heights(heights_lake=erhai_wse, markersize=7, y_var='height_weighted_mean', 
                  uncertain_var='height_weighted_dev', date_ticks=dates_ticks, h_range=[1965.2, 1966.7], ax=ax[2])
# ax[2].text(0.05, 0.1, f"Range: {max_range_erhai:.2f}m", 
#                 transform=ax[2].transAxes, fontdict=font_config)
# ax[2].set_title('Erhai', fontsize=12, y=0.83, fontweight='bold')
print(round(max_range_erhai, 2)) 
```
